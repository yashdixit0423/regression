```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn import linear_model
import math

df = pd.read_csv("test_scores.csv")

df
```

```python
X = df[['math', 'cs']].values
y = (df['math'] + df['cs']) / 2  
print(X)
print(y)# Create a synthetic target variable (average score)
```

```python
# Normalize features for gradient descent scaling the features so that each value has a mean =0 and standard deviattion =1, to vaoid bias and improve scaling
X = (X - X.mean(axis=0)) / X.std(axis=0)
```

```python
# Add intercept term
X_b = np.c_[np.ones((X.shape[0], 1)), X]  # Add a column of ones to X to account for the bias in the linear regression model
```

```python
# Gradient Descent Parameters
learning_rate = 0.01
n_iterations = 1000
m = len(y)
```

```python
# Initialize weights
theta = np.random.randn(X_b.shape[1])  # Random initialization
```

```python
for iteration in range(n_iterations):
    gradients = 2/m * X_b.T.dot(X_b.dot(theta) - y)  # Compute gradients X_b.dot(theta) =predicted values,y =actual target m=len of trainingexample
    theta -= learning_rate * gradients  # Update weights
```

```python
predictions = X_b.dot(theta)

# Display the learned parameters
print("Learned parameters (weights):", theta)
```

```python
# Example predictions for the original data
for i in range(len(df)):
    predicted_score = X_b[i].dot(theta)
    print(f'Predicted average score for {df["name"][i]}: {predicted_score:.2f}')
```
