```python
from sklearn import linear_model
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
df = pd.read_csv('Salary_dataset.csv')
df.tail()
```

```python
%matplotlib inline
plt.xlabel('Experience (in Years)')
plt.ylabel('Salary (in USD)')
plt.scatter(df.YearsExperience, df.Salary, color='red', marker='+')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['YearsExperience']], df.Salary)
```

### y=mx+c is took as the equation 
### where y is Years of Experience, 
### x is Salary, 
### m is coefficient of regression and c is intercept of y

```python
reg.predict([[33]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
p=reg.predict(df[['YearsExperience']])
p
```

### building visuals for prediction line which is the best fit curve for the datapoints and is represented in blue color here

```python
plt.scatter(df['YearsExperience'], df['Salary'], color = 'red', marker = '+')
plt.plot(df['YearsExperience'], p, color = 'blue')
plt.title("Model's interpretation of data vs actual data")
plt.ylabel('Salary (in USD)')
plt.xlabel('Experience (in years)')
plt.show()
```
